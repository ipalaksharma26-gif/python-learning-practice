```python
#University grade calculator
marks = int(input("Enter Marks"))
if marks >= 90:
    grade, remark = "O", "Outstanding"
elif marks >= 75:
    grade, remark ="A+", "Excellent"
elif marks >= 60:
    grade, marks ="A", "Very Good"
elif marks >= 50:
    grade, marks ="B", "Good"
elif marks >= 40:
    grade, marks ="C", "Average - Pass"
else:
    grade , remark ="F" , "Fail - Reppear"
print(f"Marks : {marks}")
print(f"Grade : {grade} ({remark}")
```

```python
# Income Tax Slab (India — simplified)
income = int(input("enter income"))
if income <= 300000:
    tax = 0
elif income <= 600000:
    tax = (income - 300000) * 0.05
elif income <= 900000:
    tax = 15000 + (income - 600000) * 0.10
elif income <= 1200000:
    tax = 45000 + (income - 900000) * 0.15
else:
    tax = 90000 + (income - 1200000) * 0.20
print(f"Income : Rs.{income:,}")
print(f"Tax  : Rs.{tax:,}")
```

```python
#write a program to accept thw basic salary from user and calculate the net salary and pf
basic = int(input("Enter your basic salary"))
da=basic*0.5
ta=basic*0.08
hra=basic*0.13
pf=basic*0.12
nets=basic+da+ta+hra+pf
print ("Net Salary is:",nets,"\n Pf Amount:" ,pf+pf)
```

```python
# Bank Loan Eligibility — nested conditions
age =  int(input("Enter your age"))
salary =  int(input("Enter your salary"))
credit_score = 720
existing_loans = 1
if age >= 21:
    if salary >= 30000:
        if credit_score >= 700:
            if existing_loans < 3:
                print("Loan APPROVED!")
                loan_amt = salary * 24
                print(f"Eligible amount: Rs.{loan_amt:,}")
            else:
                print("Loan REJECTED: Too many existing loans.")
        else:
            print("Loan REJECTED: Low credit score.")
    else:
        print("Loan REJECTED: Insufficient salary.")
else:
    print("Loan REJECTED: Age below 21.")
```
