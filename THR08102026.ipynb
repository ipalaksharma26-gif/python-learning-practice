```python
nth=int(input("Enter nth number"))
son=0
soe=0
soo=0
i=1
while(i<=nth):
    son=son+i
    if(i%2==0):
        soe=soe+i
    else:
        soo=soo+i
    i=i+1
print("son =",son,"/nsoe=",soe,"\nsoo=",soo)
```

```python
n=int(input("enter any number"))
fact=1
i=1
while(i<=n):
    fact=fact*i
    i=i+1
print("Factorial=",fact)
```

```python
n=int(input("enter the value of n"))
r=int(input("enter the value of r"))
nr=(n-r)
fn=1
fr=1
fnr=1
i=1
while(i<=n):
    fn=fn*i
    i=i+1
j=1
while(j<=r):
    fr=fr*j
    j=j+1
k=1
while(k<=nr):
    fnr=fnr*k
    k=k+1
nPr=fn/fnr
nCr=fn/((fnr)*(fr))
print("Permutation is:",nPr,"\nCombination is:",nCr)
```

```python
#wap to accept any digit number from user and check weather the number is palindrome?
n=int(input("enter any digit number")) # 0121
temp=n
rev=0
while(n!=0):
    r=n%10
    n=n//10
    rev=rev*10+r
if(temp==rev):
    print("Palindrome")
else:
    print("Not Palindrome")
```
