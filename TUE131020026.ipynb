```python
#Reverse a string using for loop
text=input("Enter String")
rev=""
for ch in text:
    rev=ch+rev
print("Reverse String is:",rev)
```

```python
#count vowel in a string
text=input("Enter String")
c=0
for ch in text.lower():
    if ch in"aeiou":
        c=c+1
print("No of Vowels is:",c)
```

```python
#check input number is prime or not
n=int(input("enter any number"))
is_prime=n>=2
for i in range(2,int(n**0.5)+1):
    if n%i==0:
        is_prime=False
        break
if is_prime:
    print("Prime Number")
else:
    print("Not Prime Number")
```
